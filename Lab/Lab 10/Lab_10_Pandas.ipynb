##Pandas Series

A **pandas Series** is one-dimensional labeled array containing data of the same type (integers, strings, floating point numbers, Python objects, etc.)

The ROW labels are often referred to as *index*.  The default index starts at 0 and counts up.


````python
# Example of creating Pandas series from numpy array
series1 = pd.Series(range(5,30,5))
series1
````

````python
# Creating Pandas series with a custom index
series2 = pd.Series(range(5,30,5), index = ['a', 'b', 'c', 'd', 'e'])
series2
````

````python
# Now series2 is like a dictionary, where we can get values associated with the row index.
series2['c']
````

````python
# Pandas series have functions similar to a numpy array
series2.mean()
````

````python
# Slicing series supports slicing like numpy array
series2[4:0:-1]
````

## Filtering
DataFrames can be filtered to show some records and/or some columns in a variety of ways.

````python
# get 5 RANDOM records from the data frame
# (try running this cell a couple of times!)
zoo.sample(5)
````

````python
# the results of filtering can be stored in a new data frame
# (and the columns are arranged in the order we request)
five_random_animals = zoo.sample(5)[['animal', 'name']]
five_random_animals
````

**Filtering by content**

So far we've filtered our dataframe to specific columns or range of records. Let's say we want to see a list of only the zoo animals that are injured, or only carnivores.

To filter in this way, we first create a pandas series filled with boolean values (True/False) depending on a conditional expression for the criteria. That series is then used to "index" into the dataframe and select just the records that meet the criteria.

````python
# create a series that is True when the animal's 'injured' column is 'Y'
is_injured = (zoo['injured'] == 'Y')
is_injured
````

````python
# Then use that True/False series to filter the zoo dataframe
zoo[is_injured]
````

````python
# Now let's find all the carnivores, but this time
# we'll filter it all in one line, like this:
carnivores = zoo[zoo['diet']=='carnivore']
carnivores
````

````python
# To find out how many carnivores there are, we can do:
len(carnivores)
````

````python
# We can combine boolean True false arrays using a special AND (&) operator
is_injured = zoo['injured']=='Y'
is_carnivore = zoo['diet']=='carnivore'
zoo[is_injured & is_carnivore]
````

````python
# There is also a special OR (|) operator
zoo[is_injured | is_carnivore]
````

````python
# And there's a special NOT (~) operator
# which animals are injured and NOT carnivores?
zoo[is_injured & (~ is_carnivore)]
````

````python
#We can use other relational operations to make our boolean series for filtering
is_very_thirsty = zoo['water_need'] > 500
zoo[is_very_thirsty]

````

````python
# Pandas has special support for handling dates/times
# Which animals arrived on or after January 5, 2017?
is_recent_arrival = zoo['arrival_date'] >= pd.Timestamp(2017,1,5)
zoo[is_recent_arrival]
````

````python
# Which animals arrived in April (month 4) of ANY year?
is_april_arrival = zoo['arrival_date'].dt.month == 4
zoo[is_april_arrival]
````

````python
# Pandas also has special support for applying string functions
# to a whole column of data

# Which animals have an 's' (or 'S') in their name?

lowerCaseNameColumn = zoo['name'].str.lower()
zoo[lowerCaseNameColumn.str.contains('s')]
````

## Grouping in Pandas
Instead of aggregating over ALL of the animals, what if we want to know the mean water need of each type of animal?

Here's how pandas performs "segmentation" (grouping and aggregation) based on the column values.

````python
# Use groupby to group the values in the 'animal' column and then find the mean
zoo.groupby('animal').mean()
````

*Note that we don't care about the mean for the zoo_id column, since it doesn't make sense to average ID #s!*

````python
# To avoid this issue...

# Use groupby to group the values in the 'animal' column and only display the mean of the water_need column

zoo.groupby('animal').mean()[['water_need']]
````

````python
# How many animals are there of each diet type?

# it doesn't really matter which column we look at,
# since all columns will have the same count.
# we'll arbitrarily choose the animal column
countsByDiet = zoo.groupby('diet').count()['animal']

#then we'll sort it by the animal count values
countsByDiet.sort_values()
````

## Visualizing data with pandas

Pandas doesn't have provide of the bells and whistles that **matplotlib** provides for creating so many different custom plots, BUT it makes it easy to create a few simple graphs/plot types from your dataframes.

````python
#For a small data frame, we can call df.plot(...), and provide
# x = the column name we'd like to plot on the x-axis
# y = the column name we'd like to display on the y-axis
# kind = the type of plot/chart ('barh' for horizontal bar chart)
# figsize = (width,height) dimensions for the output graph
zoo.plot(x='name', y='water_need', kind='barh', figsize=(10,6))
````

````python
# It might be a better visualization if we sorted it first.
zoo.sort_values('water_need').plot(x='name', y='water_need', kind='barh', figsize=(10,6))
````

````python
# We can also combine plotting with groupby, to visualize info about subsets of data
#
# AND there are some options for the 'kind' of plot, such as:
#  * line (default), bar (vertical), barh, hist, box, density, scatter
# For more info, see: https://pandas.pydata.org/pandas-docs/stable/user_guide/visualization.html

#Which type of animals requires the most water on average?

#In this case, pandas can guess the x & y data without us specifying
#However, we do need to add a y-axis label so the user knows what's being graphed
meanWaterNeedByAnimal = zoo.groupby('animal').mean()['water_need'].sort_values()
meanWaterNeedByAnimal.plot(kind='bar',ylabel='average water needed')
````

````python
# Instead of looking at just the **average** water need,
# we can try to get a sense of shape of the data using a boxplot
# which shows average values but ALSO the distribution of data
zoo.boxplot(column='water_need', by='animal', figsize=(8,5))
````

*It appears the widest variation in water requirements is among the lions.*

---


````python
# What is the composition of our zoo?
# What fraction of animals of each type do we have?
zoo.groupby('animal').count()['zoo_id'].plot(kind='pie')
````

````python
# However, there is also a handy value_counts() method that does groupby counting
# and sorting by those counts all in one command.

zoo['animal'].value_counts().plot(kind='pie')
````

*Notice how now the pie slices are arranged in strictly decreasing order by size, going counter-clockwise from **zebra**, which is what the zoo has the most of?*

---


````python
# How about looking at how many animals the zoo acquired each year?

# Pandas provides a special way to extract the year from the date column
# and we can group by that.

arrivalsByYear = zoo.groupby(zoo.arrival_date.dt.year).count()[['zoo_id']]
arrivalsByYear.columns = ['arrivals']
arrivalsByYear
````

````python
#then we can plot our grouped data frame, using circular markers with dotted lines between!
arrivalsByYear.plot(kind='line',figsize=(12,4), style=':o',ylabel='# of animals')

#and we can use matplotlib to tweak the plot settings a bit
import matplotlib.pyplot as plt
plt.ylim(0,5)  # y-axis should go from 0 to 5
plt.xticks(range(1999,2020)) #show all years on x-axis
plt.show()
````

Neat graph!  But wait... note that this plot is misleading because it is drawing lines betewen the dots, but it isn't visually displaying any of the years that ZERO animals arrived!

So instead of grouping & plotting this way, it might be better to start over and use a histogram...

````python
arrivalYears = zoo['arrival_date'].dt.year
arrivalYears.plot(kind='hist',bins=range(1999,2021),color='yellow',edgecolor='black',figsize=(12,4))
plt.xticks(range(1999,2020)) #show all years on x-axis
plt.show()
````
